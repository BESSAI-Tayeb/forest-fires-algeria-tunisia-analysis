# Merge All Datasets

This notebook merges all processed datasets from the results folder into one unified dataset.


```python
import pandas as pd
import geopandas as gpd
import rasterio
from pathlib import Path

# Load all datasets from results
results_path = Path('../results/')

# Load tabular data
soil_clean = pd.read_parquet(results_path / 'soil_features_cleaned.parquet')
soil_engineered = pd.read_parquet(results_path / 'soil_features_engineered.parquet')
fire_merged = pd.read_parquet(results_path / 'merged_fire_data.parquet')

# For spatial data, we need to decide on merging strategy
# Since datasets have different geometries, create separate GeoDataFrames

# Fire data is already points
fire_gdf = gpd.GeoDataFrame(
    fire_merged,
    geometry=gpd.points_from_xy(fire_merged['longitude'], fire_merged['latitude']),
    crs='EPSG:4326'
)

# Soil data might need spatial join with units
# For now, save as separate files or create a unified structure

# Save merged datasets
fire_gdf.to_parquet(results_path / 'final_merged_fire.parquet')
soil_clean.to_parquet(results_path / 'final_merged_soil.parquet')

print('Datasets merged and saved')
```
