# Merge All Datasets

This notebook merges all processed datasets from the results folder into one unified dataset.


```python
import pandas as pd
import os

# Define the results folder path
results_folder = '../results'

# Load the base fires dataset
fires_df = pd.read_csv(os.path.join(results_folder, 'fires.csv'))

# Load additional feature datasets
climate_df = pd.read_csv(os.path.join(results_folder, 'fires_with_climate.csv'))
elevation_df = pd.read_csv(os.path.join(results_folder, 'fires_with_elevation_features.csv'))
soil_df = pd.read_csv(os.path.join(results_folder, 'fires_with_soil_features.csv'))

# Merge datasets on longitude and latitude
# Start with fires_df and merge climate
merged_df = fires_df.merge(climate_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge elevation features
merged_df = merged_df.merge(elevation_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge soil features
merged_df = merged_df.merge(soil_df, on=['longitude', 'latitude', 'class'], how='left')

# Optionally, load and merge other datasets if needed (e.g., engineered features)
# engineered_df = pd.read_csv(os.path.join(results_folder, 'fires_engineered_features.csv'))
# merged_df = merged_df.merge(engineered_df, on=['longitude', 'latitude', 'class'], how='left')

# Save the merged dataset
merged_df.to_csv(os.path.join(results_folder, 'fires_merged_all_features.csv'), index=False)

print("Merged dataset saved to fires_merged_all_features.csv")
print(f"Shape of merged dataset: {merged_df.shape}")
print(merged_df.head())
```
