# Merge All Datasets

This notebook merges all processed datasets from the results folder into unified datasets for supervised and unsupervised learning.

WARNING: this cell is for supervised

```python
# Add project root to path and import config
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent.parent))
from config import (
    FIRES_WITH_ARTIFICIAL_SUPERVISED,
    FIRES_WITH_CLIMATE_SUPERVISED, FIRES_WITH_ELEVATION_SUPERVISED, 
    FIRES_WITH_SOIL_SUPERVISED, FIRES_MERGED_SUPERVISED
)

import pandas as pd

print("="*80)
print("MERGING SUPERVISED DATASETS")
print("="*80)

# Load the base fires dataset for supervised learning
fires_sup_df = pd.read_csv(FIRES_WITH_ARTIFICIAL_SUPERVISED)
print(f"Base fires dataset (supervised): {fires_sup_df.shape}")

# Load additional feature datasets for supervised
climate_sup_df = pd.read_csv(FIRES_WITH_CLIMATE_SUPERVISED)
print(f"Climate features: {climate_sup_df.shape}")

elevation_sup_df = pd.read_csv(FIRES_WITH_ELEVATION_SUPERVISED)
print(f"Elevation features: {elevation_sup_df.shape}")

soil_sup_df = pd.read_csv(FIRES_WITH_SOIL_SUPERVISED)
print(f"Soil features: {soil_sup_df.shape}")

# Merge datasets on longitude and latitude
# Start with fires_df and merge climate
merged_sup_df = fires_sup_df.merge(climate_sup_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge elevation features
merged_sup_df = merged_sup_df.merge(elevation_sup_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge soil features
merged_sup_df = merged_sup_df.merge(soil_sup_df, on=['longitude', 'latitude', 'class'], how='left')

# Save the merged dataset
merged_sup_df.to_csv(FIRES_MERGED_SUPERVISED, index=False)

print(f"\n✓ Merged supervised dataset saved to {FIRES_MERGED_SUPERVISED}")
print(f"  Shape: {merged_sup_df.shape}")
print(f"\nClass distribution:")
print(merged_sup_df['class'].value_counts().sort_index())
print(f"\nSample (first 3 rows):")
display(merged_sup_df.head(3))
```
