# Merge All Datasets

This notebook merges all processed datasets from the results folder into one unified dataset.


```python
# Add project root to path and import config
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent.parent))
from config import (
    FIRES_RESULT_CSV, FIRES_WITH_CLIMATE, FIRES_WITH_ELEVATION, 
    FIRES_WITH_SOIL, FIRES_MERGED
)

import pandas as pd

# Load the base fires dataset
fires_df = pd.read_csv(FIRES_RESULT_CSV)

# Load additional feature datasets
climate_df = pd.read_csv(FIRES_WITH_CLIMATE)
elevation_df = pd.read_csv(FIRES_WITH_ELEVATION)
soil_df = pd.read_csv(FIRES_WITH_SOIL)

# Merge datasets on longitude and latitude
# Start with fires_df and merge climate
merged_df = fires_df.merge(climate_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge elevation features
merged_df = merged_df.merge(elevation_df, on=['longitude', 'latitude', 'class'], how='left')

# Merge soil features
merged_df = merged_df.merge(soil_df, on=['longitude', 'latitude', 'class'], how='left')

# Save the merged dataset
merged_df.to_csv(FIRES_MERGED, index=False)

print(f"Merged dataset saved to {FIRES_MERGED}")
print(f"Shape of merged dataset: {merged_df.shape}")
print(merged_df.head())
```
